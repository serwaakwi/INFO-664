## Data types in Python

Values in Python can be a number of different data types, such as **int, float, string, bool**

```python
print(type(5)) # type function returns data type of specific parameter - this will return "int" for integer
print(type(5.5)) # float - decimal point number
print(type("5")) # str (string) - text
print(type(True)) # bool (boolean) - true or false
```

## Comparison Operators

Used to compare data, which evaluate to **True** or **False**

```python
print(5 > 3) # "greater than" operator
print(5 == 5) # "equal to" operator
print(5 != 5) # inequality operator
print(5 <= 3) # "less than or equal to" operator 
```

5 **is** greater than 3, hence **True**.
5 **is** equal to 5, hence **True**.
5 **is not** inequal to 5, hence **False**.
5 **is not** less than or equal to 3, hence **False**.

## Arithmetic Operators

These are math operators, used to perform math operations on data

```python
print(7+2) # addition
print(7-2) # subtraction
print(7*2) # multiplication 
print(7/2) # "true" division -- will return a decimal/float in this case
print(7//2) # "floor" division -- will round down result
print(7%2) # modulo, will return remainder of division of these two numbers
print(7**2) # exponent
```

## Variable Declaration

Variables in Python can be declared using the "=" symbol. The word coming before the equal symbol (on the left side) is the **name** of the variable, and its **value** will be on the other side of the equal symbol.

The variable will store this information throughout the notebook.

```python
name = "Akua"
age = 24
temperature = 0.5
```

```python
print(name)
```

### Naming conventions with variables

- your variable name should represent what it is holding
- names ***cannot*** start with a number
- names ***cannot*** have a hyphen in them
- names ***can*** have underscores in them
    - place_of_birth is a valid variable name using *snake case*
- names ***cannot*** be special words in the Python language like "if" and "for" (which are generally bad variable names anyways)

## Converting data types

Data can be converted from one type to another using conversion functions:

```python
pi = 3.14159
print(pi)

pi_as_string = str(pi) # str converts to string
print(pi_as_string)
print(type(pi_as_string))

pi_as_float_again = float(pi_as_string) #float converts to float (duh)
print(pi_as_float_again)
print(type(pi_as_float_again))
```

```python
print(pi_as_string>3) #cannot compare two different data types
```

```python
print(pi_as_float_again>3)
```

More on the quotes:

```python
sample_text = "i'm having a great day!" # double quotes are being used here because an apostrophe is involved
print(sample_text)
wrong_sample_text = 'i'm having a great day!' # interpreted as "i" instead of "i'm"
print(wrong_sample_text)
```

## Combining and concatenating strings

Strings can be combined a number of ways. One is to use the + operator!

```python
dog_name = "Fido"
dog_breed = "Labrador"
dog_age = 4

my_dog_string = "My dog's name is " + dog_name + " and he is " + str(dog_age) # must convert int to str

print(my_dog_string)
```

We can also use [f-strings](https://www.w3schools.com/python/python_string_formatting.asp)!

```python
my_dog_string = f"My dog's name is {dog_name}. He is {dog_age}, and he is a {dog_breed}." 
# f is a special indicator... this is a special string... note that i didn't convert the integer...
print(my_dog_string)
```

## Converting string case

There are a bunch of built-in methods for converting strings to upper, lower, capitalized, and title case

```python
print(dog_name.upper()) 
print(dog_name.lower())
print(my_dog_string.title())
print(my_dog_string.capitalize())
```

## Searching, counting, extracting

There are some string methods in Python that deal with searching for substrings, etc. One of those is the keyword **in**.

```python
print("cat" in my_dog_string) #will return true or false if the word "cat" can be found in that string
```

```python
print("dog" in my_dog_string)
```

```python
my_dog_string = my_dog_string + " I really love dogs and dogs are my best friends."
print(my_dog_string)
```

```python
dog_count = my_dog_string.count("dog")
print(f"The word 'dog' appears {dog_count} times.")
```

## Cleaning up text with strip()

```python
messy_string = "     I have a bunch of unnecessary white space before and after me!!!     "
print(messy_string)
print(messy_string.strip())
```

You can find the length of a string using the `len()`function, which returns the length of characters

```python
print(len(messy_string))
clean_string = messy_string.strip()
print(len(clean_string))
```

## Text replacement!
`.replace(old, new)` searches for one section of a string and replaces it with another!

```python
mood_string = "I am having a terrible time programming in Python"
print(mood_string)
positive_mood_string = mood_string.replace("terrible","wonderful")
print(positive_mood_string)
```

## Parsing comma-delimited data
Python has methods for parsing data out of strings by delimiting characters. For instance, commas are often used to separate values in CSV-style datasets. The `split(",")` method will separate a string based on locations of commas.


```python
csv_data = "Nirvana, Smells Like Teen Spirit, 5:01, 1991"
artist, song, length, year = csv_data.split(",")
print(f"My favorite band is {artist}, and their best song is {song}, which came out in {year}, and is {length}")
```

## Slicing a string
We can extract a certain range of characters from a string using the slicing approach:

```python
my_text = "saxophone"
print(my_text[2:5]) #finds characters 2 through 5
print(my_text[0]) #shows character zero
print(len(my_text)) #shows length of text
print(my_text[len(my_text)-1]) #prints the last character
print(my_text[-1]) #prints the last character by counting from the end
print(my_text[-4:]) #-4 refers to 4 characters from the end. : prints range to the end
```
