3. Use the MET API to assemble a new list of dictionaries containing only items that relate to a specific animal, person, place. This doesn’t have to be comprehensive, as you may end up with thousands of items, so make a list that has at least 5 items in it.

```python
import json
import requests
from time import sleep

base_url = "https://collectionapi.metmuseum.org"
search_path = "/public/collection/v1.1/search"
object_path = object_path = "/public/collection/v1/objects/"
search_query = "?q=elephant"
search_url = f'{base_url}{search_path}{search_query}'

search_response = requests.get(search_url)

elephant_objects = search_response.json()

print(elephant_objects.keys())
print(elephant_objects["objectIDs"])

elephant_object_ids = elephant_objects["objectIDs"]

elephant_objects_details = []

for elephant_id in elephant_object_ids:
    url=f'{base_url}{object_path}{elephant_id}'
    response_object = requests.get(url) #making the request to the server
    elephant_objects_details.append(response_object.json())
    sleep(0.5)
print(len(elephant_objects_details))

```
