Below you will see a `for` loop that will iterate through every item in the list. Notice the syntax: the indented portion is the portion of the for loop that is repeated

```python
print(routine)
for step in routine: #extract each element in routine one by one, referring to it as a "step"
    print(step)
```

A `for` loop consists of several elements:
- the `for` keyword, which indicates we are declaring a looping structure
- an incremented variable (step), which will represent each element in the list -- it can be called whatever you want. often it is called "i"  as a generic term, but it often makes sense to refer to whatever a specific entry is in your list
- the `in` keyword, followed by a list or collection (anything that can be iterated through)
- the **:** (colon) character, which tells python that the looping lines of code are to follow. The lines below a colon will be indented one level further than we currently are
- and finally, the funcitons to loop. these are functions that are repeated for every element in the list

```python
fruit = "banana"
for i in fruit:
    print(i)
```

```python
fruit = "banana"
for i in fruit:
    print(i.upper()) #can modify using functions 
```

# Working with CSV data

CSV is one of the most common dataset formats, it's basically a spreadsheet with rows and columns
Each row is a new line, each column is separated by a comma
CSVs usually have a "header" row that is in the first row of the document
Each subsequent row is the actual data

Let's have a look at the MoMA dataset moma_artists.csv

To work with CSV data in python we need to import the CSV library, by typing `import csv` before attempting to use any csv functions

We can open a file using the following syntax:

`with open(file_name, options) as file:`

```python
import csv

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)  #initializing an object that interprets csv data
    #the below line will extract all the headings
    header = next(reader) #tells the reader to read the first line and move to the next
    rows = list(reader) #takes every other row in the dataset and makes it into a list
print(header)
print(rows[0]) #first element in list
```

We can also generate a list of dictionaries that use the header to establish a set of shared keys, assigning values for each row based on those keys/headers.

We can do that by using the csv method `DictReader(file)`

```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) #this will interpret the csv as a list of dictionaries
    data = list(reader) # store the csv data as a list of dictionaries we can access later

print(data[-1])
```

```python
# filtering by nationality: only Swedish artists
swedish_artists = [] #creating an empty list before running for loop
for artist in data: #for each artist in my data
    if artist["Nationality"] == "Swedish":
        swedish_artists.append(artist)
print(len(swedish_artists))
```
