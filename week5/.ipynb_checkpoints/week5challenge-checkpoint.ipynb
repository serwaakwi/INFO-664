```python
#goal: iterate through list and eliminate duplicates. if "not" works this way
artist_year = []
for artist in data:
  artist_year.append(artist["BeginDate"])
print(artist_year[0])
```

```python
unique_artist_year = []
for year in artist_year:
  if year not in unique_artist_year:
    unique_artist_year.append(year)
print(len(unique_artist_year))    #ideally is the amount of years represented in the data
```
